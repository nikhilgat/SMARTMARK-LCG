```python
from ultralytics import YOLO

model = YOLO("yolo26x.pt")  # load a pretrained YOLO26s model
results = model("room.jpg")  # run inference
```

```python
from PIL import Image

Image.fromarray(results[0].plot()[..., ::-1])  # plot() returns BGR; flip to RGB for display
```

```python
segmentation_model = YOLO("yolo26x-seg.pt")  # load a pretrained YOLOv8 segmentation model

results = segmentation_model("room.jpg")  # run segmentation inference
```

```python
from PIL import Image

Image.fromarray(results[0].plot()[..., ::-1])  # plot() returns BGR; flip to RGB for display
```
